```python
from ollama import chat
import ipywidgets as widgets
from IPython.display import display, Markdown

def call_llm(system_prompt, context, user_prompt, model="mistral:7b-instruct",
             temperature=0.2, top_p=0.9):
    messages = [
        {"role": "system", "content": system_prompt.strip()},
        {"role": "system", "content": f"OFERTA (KONTEKST):\n{context.strip()}"},
        {"role": "user", "content": user_prompt.strip()},
    ]
    resp = chat(model=model, messages=messages, options={"temperature": float(temperature), "top_p": float(top_p)})
    return resp["message"]["content"]

```

```python
system_w = widgets.Textarea(
    value="Jestes asystentem internetowego sklepu sprzedającego elektronikę, nie odpowiadaj na pytania związane z czymkolwiek innym, niz sprzedawane produkty.",
    description="System:",
    layout=widgets.Layout(width="100%", height="120px")
)

context_w = widgets.Textarea(
    value="komputer stacjonarny - procesor i5, grafika rtx3070, 24gb ramu ddr4,"
"nawilzacz powietrza - wydajnosc 24mkw/h,"
"laptop gamingowy - procesor i7, grafika rtx5070, 32gb ramu ddr4,",
    description="Kontekst:",
    layout=widgets.Layout(width="100%", height="140px")
)

user_w = widgets.Textarea(
    value="Co to jest sztuczna inteligencja?",
    description="Pytanie:",
    layout=widgets.Layout(width="100%", height="100px")
)

temperature_w = widgets.FloatSlider(
    value=0.7,
    min=0.0,
    max=1.2,
    step=0.1,
    description="Temp:"
)

top_p_w = widgets.FloatSlider(
    value=0.9,
    min=0.1,
    max=1.0,
    step=0.05,
    description="Top-p:"
)

run_btn = widgets.Button(
    description="🚀 Zapytaj bota",
    button_style="success"
)

out = widgets.Output()

display(system_w)
display(context_w)
display(user_w)
display(temperature_w)
display(top_p_w)
display(run_btn)
display(out)
```

```python
def on_click(b):

    out.clear_output()

    with out:

        display(Markdown("## 🧠 Prompt wysłany do modelu"))

        full_prompt = f"""
SYSTEM:
{system_w.value}

KONTEKST:
{context_w.value}

PYTANIE:
{user_w.value}
"""

        display(Markdown(full_prompt))

        display(Markdown("---"))
        display(Markdown("## 🤖 Odpowiedź:"))

        answer = call_llm(
            system_w.value,
            context_w.value,
            user_w.value,
            temperature=temperature_w.value,
            top_p=top_p_w.value
        )

        display(Markdown(answer))

run_btn.on_click(on_click)
```
