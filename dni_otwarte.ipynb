```python
#!pip install ollama ipywidgets
```

```python
from ollama import chat
import ipywidgets as widgets
from IPython.display import display, Markdown

def call_llm(system_prompt, context, user_prompt, model="llama3.2:3b", # Wybór modelu sztucznej inteligencji 'mistral:7b-instruct'
             temperature=0.2, top_p=0.9):
    messages = [
        {"role": "system", "content": system_prompt.strip()},
        {"role": "system", "content": f"OFERTA (KONTEKST):\n{context.strip()}"},
        {"role": "user", "content": user_prompt.strip()},
    ]
    resp = chat(model=model, messages=messages, options={"temperature": float(temperature), "top_p": float(top_p)})
    return resp["message"]["content"]

```

---

## 🧠 Jak działa ten chatbot?

Chatbot składa się z trzech elementów:

### 1️⃣ SYSTEM
To instrukcja dla AI – kim ma być i jak ma się zachowywać.

### 2️⃣ KONTEKST
To wiedza, którą mu przekazujemy.

### 3️⃣ PYTANIE
To to, co użytkownik chce wiedzieć.

---

## 🎛️ Parametry sterujące

- **Temperature** – jak bardzo kreatywna ma być odpowiedź
- **Top-p** – jak szeroki zakres możliwych słów może wybrać model

---

Zmodyfikuj pola poniżej i zobacz, co się stanie!

```python
system_w = widgets.Textarea(
    value="Jesteś interaktywnym przewodnikiem po świecie sztucznej inteligencji."
          "Mówisz prostym językiem dla ucznia 8 klasy."
          "Odpowiadasz jasno, w 4–6 zdaniach."
          "Na końcu każdej odpowiedzi zadajesz jedno ciekawe pytanie zachęcające do dalszej rozmowy."
          "Nie używaj skomplikowanego żargonu technicznego.",
    description="System:",
    layout=widgets.Layout(width="100%", height="120px")
)

context_w = widgets.Textarea(
    value="Znajdujemy się na dniach otwartych technikum informatycznego."
          ""
          "Szkoła oferuje kierunki:"
          "- Programista"
          "- Technik informatyk"
          "- Technik robotyki"
          ""
          "Szkoła posiada:"
          "- pracownię AI"
          "- drukarki 3D"
          "- laboratorium sieci komputerowych"
          "- koło tworzenia gier"
          ""
          "Uczniowie uczą się Pythona, budują roboty i tworzą własne aplikacje.",
    description="Kontekst:",
    layout=widgets.Layout(width="100%", height="140px")
)

user_w = widgets.Textarea(
    value="Czym jest sztuczna inteligencja i czy w naszej szkole mogę się jej nauczyć?",
    description="Pytanie:",
    layout=widgets.Layout(width="100%", height="100px")
)

temperature_w = widgets.FloatSlider(
    value=0.7,
    min=0.0,
    max=1.2,
    step=0.1,
    description="Temp:"
)

top_p_w = widgets.FloatSlider(
    value=0.9,
    min=0.1,
    max=1.0,
    step=0.05,
    description="Top-p:"
)

run_btn = widgets.Button(
    description="🚀 Zapytaj bota",
    button_style="success"
)

out = widgets.Output()

display(system_w)
display(context_w)
display(user_w)
display(temperature_w)
display(top_p_w)
display(run_btn)
display(out)
```

---

## 🔬 Co właśnie testujesz?

Za każdym razem wysyłamy do modelu:

- instrukcję (SYSTEM)
- dane wejściowe (KONTEKST)
- pytanie użytkownika

Model generuje odpowiedź na podstawie tych trzech elementów.

Spróbuj:
- zmienić styl wypowiedzi
- ograniczyć długość odpowiedzi
- dodać zakazy
- zmienić temperaturę

```python
def on_click(b):

    out.clear_output()

    with out:

        display(Markdown("## 🧠 Prompt wysłany do modelu"))

        full_prompt = f"""
SYSTEM:
{system_w.value}

KONTEKST:
{context_w.value}

PYTANIE:
{user_w.value}
"""

        display(Markdown(full_prompt))

        display(Markdown("---"))
        display(Markdown("## 🤖 Odpowiedź:"))

        answer = call_llm(
            system_w.value,
            context_w.value,
            user_w.value,
            temperature=temperature_w.value,
            top_p=top_p_w.value
        )

        display(Markdown(answer))

run_btn.on_click(on_click)
```
